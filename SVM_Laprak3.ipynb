### 1. Import library

```python
#Tashfia Wahdah Fuady_202431013
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### 2. Membaca dan memahami dataset

```python
#Tashfia Wahdah Fuady_202431013
df = pd.read_csv("data_breast cancer.csv")

print("5 Data Pertama:")
display(df.head())

print("\nInformasi Dataset:")
df.info()

print("\nStatistik Deskriptif:")
display(df.describe())
```

### 3. Membersihkan data

```python
#Tashfia Wahdah Fuady_202431013
# Menghapus kolom ID dan kolom kosong
df = df.drop(columns=["id", "Unnamed: 32"])

print("Ukuran Dataset Setelah Dibersihkan:", df.shape)

print("\nJumlah Missing Value:")
print(df.isna().sum())

# Menghapus baris jika masih terdapat missing value
df = df.dropna()

print("\nUkuran Dataset Akhir:", df.shape)
```

### 4. Memeriksa target

```python
#Tashfia Wahdah Fuady_202431013
print("Nilai Unik Diagnosis:")
print(df["diagnosis"].unique())

print("\nJumlah Setiap Kelas:")
print(df["diagnosis"].value_counts())

# B = Benign (tumor jinak)
# M = Malignant (tumor ganas)
```

### 5. Mengubah target menjadi angka

```python
#Tashfia Wahdah Fuady_202431013
le = LabelEncoder()

df["diagnosis"] = le.fit_transform(df["diagnosis"])

print("Pemetaan Kelas:")
for i, kelas in enumerate(le.classes_):
    print(kelas, "=", i)

print("\nData Setelah Encoding:")
display(df.head())
```

### 6. Memisahkan fitur dan target

```python
#Tashfia Wahdah Fuady_202431013
X = df.drop(columns=["diagnosis"])
y = df["diagnosis"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=0,
    stratify=y
)

print("Jumlah Data Latih:", X_train.shape[0])
print("Jumlah Data Uji:", X_test.shape[0])

print("\nProporsi Kelas Data Latih:")
print(y_train.value_counts(normalize=True))

print("\nProporsi Kelas Data Uji:")
print(y_test.value_counts(normalize=True))
```

### 7. Membuat pipeline SVM

```python
#Tashfia Wahdah Fuady_202431013
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("svc", SVC())
])

print(pipe)
```

### 8. Menentukan parameter GridSearchCV

```python
#Tashfia Wahdah Fuady_202431013
param_grid = {
    "svc__kernel": ["linear", "rbf", "poly"],
    "svc__C": [0.1, 1, 10, 100],
    "svc__gamma": ["scale", 0.01, 0.1, 1]
}

jumlah_kombinasi = 3 * 4 * 4

print("Jumlah Kombinasi Parameter:", jumlah_kombinasi)
```

### 9. Menjalankan GridSearchCV

```python
#Tashfia Wahdah Fuady_202431013
grid = GridSearchCV(
    pipe,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=1,
    verbose=2
)

grid.fit(X_train, y_train)

print("Parameter Terbaik:")
print(grid.best_params_)

print("\nSkor Validasi Silang Terbaik:")
print(grid.best_score_)
```

### 10. Memeriksa kandidat terbaik

```python
#Tashfia Wahdah Fuady_202431013
hasil_grid = pd.DataFrame(grid.cv_results_)

kandidat_terbaik = hasil_grid[
    hasil_grid["rank_test_score"] == 1
][["params", "mean_test_score", "std_test_score", "rank_test_score"]]

print("Kandidat dengan Peringkat Pertama:")
display(kandidat_terbaik)

print("\nModel yang Dipilih:")
print(grid.best_params_)
```

### 11. Evaluasi model

```python
#Tashfia Wahdah Fuady_202431013
best_svm = grid.best_estimator_

y_train_pred = best_svm.predict(X_train)
y_test_pred = best_svm.predict(X_test)

print("Akurasi Data Latih:", accuracy_score(y_train, y_train_pred))

print("\nConfusion Matrix Data Latih:")
display(pd.DataFrame(
    confusion_matrix(y_train, y_train_pred),
    index=["Aktual 0", "Aktual 1"],
    columns=["Prediksi 0", "Prediksi 1"]
))

print("\nClassification Report Data Latih:")
print(classification_report(
    y_train, y_train_pred,
    target_names=le.classes_
))

print("\nAkurasi Data Uji:", accuracy_score(y_test, y_test_pred))

print("\nConfusion Matrix Data Uji:")
display(pd.DataFrame(
    confusion_matrix(y_test, y_test_pred),
    index=["Aktual 0", "Aktual 1"],
    columns=["Prediksi 0", "Prediksi 1"]
))

print("\nClassification Report Data Uji:")
print(classification_report(
    y_test, y_test_pred,
    target_names=le.classes_
))
```

### 12. Analisis margin pada dua dimensi

```python
#Tashfia Wahdah Fuady_202431013
# Memilih dua fitur
fitur_2d = ["radius_mean", "texture_mean"]

X_train_2d = X_train[fitur_2d]

# Menggunakan nilai C terbaik dari GridSearchCV
C_terbaik = grid.best_params_["svc__C"]

# Standardisasi fitur
scaler_2d = StandardScaler()
X_train_scaled = scaler_2d.fit_transform(X_train_2d)

# Membuat SVM linear
svm_linear = SVC(kernel="linear", C=C_terbaik)
svm_linear.fit(X_train_scaled, y_train)

print("Nilai C:", C_terbaik)
print("Coef:", svm_linear.coef_)
print("Intercept:", svm_linear.intercept_)
print("Jumlah Support Vector:", svm_linear.n_support_)

# Menghitung margin
w = svm_linear.coef_[0]
margin = 2 / np.linalg.norm(w)

print("\nLebar Margin:", margin)
```

### 13. Uji satu data baru

```python
#Tashfia Wahdah Fuady_202431013
# Mengambil satu data dari data uji
data_baru = X_test.iloc[[0]]

# Memeriksa rentang fitur terhadap data latih
rentang = pd.DataFrame({
    "Nilai Data Baru": data_baru.iloc[0],
    "Minimum Data Latih": X_train.min(),
    "Maksimum Data Latih": X_train.max()
})

rentang["Dalam Rentang"] = (
    (rentang["Nilai Data Baru"] >= rentang["Minimum Data Latih"]) &
    (rentang["Nilai Data Baru"] <= rentang["Maksimum Data Latih"])
)

print("Pemeriksaan Rentang Fitur:")
display(rentang)

# Melakukan prediksi
prediksi = best_svm.predict(data_baru)

print("Hasil Prediksi:", le.inverse_transform(prediksi)[0])
print("Kelas Aktual:", le.inverse_transform(y_test.iloc[[0]])[0])
```
